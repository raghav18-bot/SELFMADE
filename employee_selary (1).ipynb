# Encoding the Categorical Columns

```python
data.head(1)
```

```python
data['Gender'].unique()
```

```python
data['Gender'] = data['Gender'].map({'Male':0,'Female':1})
```

```python
data['Education'].unique()
```

```python
data['Education'] = data['Education'].map({'Master':2,'Bachelor':1,'Diploma':0,'PhD':3})
```

```python
data['Department'].unique()
```

```python
data['Department'] = data['Department'].map({'Operations':0,'Marketing':1,'Sales':2,'IT':3,'HR':4})
```

```python
data['Job_Level'].unique()
```

```python
data['Job_Level'] = data['Job_Level'].map({'Junior':0, 'Lead':1, 'Mid':2, 'Senior':3, 'Manager':4})
```

```python
data['Remote_Work'].unique()
```

```python
data['Remote_Work'] = data['Remote_Work'].map({'Yes':1, 'No':0})
```

```python
data['City'].unique()
```

```python
data['City'] = data['City'].map({'Chennai':0, 'Delhi':1, 'Hyderabad':2, 'Mumbai':3})
```

```python
data.head(1)
```

# use mean and medium in missing data

```python
data['Age'].fillna(data['Age'].mean(), inplace=True)
```

```python
data['Experience_Years']= data['Experience_Years'].fillna(data['Experience_Years'].median(), inplace=True)
```

```python
data['Skill_Score']=data['Skill_Score'].fillna(data['Skill_Score'].median(), inplace=True)
```

# Store Feature Matrix In X and Response(Target) In Vector y

```python
X = data.drop(['Employee_ID','Annual_Salary_LPA'],axis=1)
y = data['Annual_Salary_LPA']
```

```python
X
```



# Splitting The Dataset into The training set and test set

```python
#!pip install scikit-learn
#!pip install xgboost
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.20,random_state=42)
```

```python
X_test
```

# Import The models

```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import GradientBoostingRegressor
from xgboost import XGBRegressor
```

# Model Training

```python
lr = LinearRegression()
lr.fit(X_train,y_train)

rf = RandomForestRegressor()
rf.fit(X_train,y_train)

xgb = GradientBoostingRegressor()
xgb.fit(X_train,y_train)

xg = XGBRegressor()
xg.fit(X_train,y_train)
```

# Prediction on Test Data

```python
y_pred1 = lr.predict(X_test)
y_pred2 = rf.predict(X_test)
y_pred3 = xgb.predict(X_test)
y_pred4 = xg.predict(X_test)
```

```python
X_test.head()
```

```python
y_test
```

# Evaluating the Algorithm

```python
from sklearn import metrics
```

```python
score1 = metrics.r2_score(y_test,y_pred1)
score2 = metrics.r2_score(y_test,y_pred2)
score3 = metrics.r2_score(y_test,y_pred3)
score4 = metrics.r2_score(y_test,y_pred4)
```

```python
print(score1,score2,score3,score4)
```

```python
final_data = pd.DataFrame({'Models':['LR','RF','GBR','XG'],
             "R2_SCORE":[score1,score2,score3,score4]})
```

```python
final_data
```

```python
sns.barplot(x=final_data['Models'], y=final_data['R2_SCORE'])
```
